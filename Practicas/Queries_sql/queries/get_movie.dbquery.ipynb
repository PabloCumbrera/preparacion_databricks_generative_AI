```sql
CREATE OR REPLACE FUNCTION cineai.app.get_movie(
  p_title STRING COMMENT 'Exact movie title'
)
RETURNS STRING
COMMENT 'Finds a movie by title. Input: exact movie title. Output: JSON with ID, title, director, genre, duration, rating, and synopsis. Failure: NOT_FOUND.'
RETURN
SELECT COALESCE(
  MAX(to_json(named_struct(
    'movie_id', movie_id,
    'title', title,
    'director', director,
    'genre', genre,
    'duration_min', duration_min,
    'age_rating', age_rating,
    'synopsis', synopsis
  ))),
  'NOT_FOUND'
)
FROM cineai.raw.movies
WHERE lower(title) = lower(trim(p_title));
```
